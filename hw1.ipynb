## Q2
Sometimes it is better to show percentage instead of real count in bar plot. Percentage helps us understand the proportion of each group. If total sample size is big, real count can be confusing.
Also, if we compare two datasets with different total size, percentage makes comparison easier and more fair.
One improvement to the plot is adding value labels on top of each bar. Also rotating x-axis labels can make it more readable if they overlap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("Salaries.csv")

# Preview data
df.head()
```

### Scatter Plot: Salary vs Years Since PhD

```python
plt.figure(figsize=(8,6))

plt.scatter(df["yrs.since.phd"], df["salary"])

plt.title("Salary vs Years Since PhD")
plt.xlabel("Years Since PhD")
plt.ylabel("Salary")

plt.show()
```
